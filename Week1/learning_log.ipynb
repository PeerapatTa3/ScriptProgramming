## 1. Set Up the Working Directory

Let's inspect the current working directory and prepare the project structure for creating the learning log.

```python
# Inspect current working directory
import os
print("Current Working Directory:", os.getcwd())
print("Contents of current directory:")
for item in os.listdir('.'):
    print(f"  - {item}")
```

## 2. Import Required Libraries

We will import the necessary Python libraries to handle file creation and path management.

```python
# Import required libraries
from pathlib import Path
import csv
from datetime import datetime
```

## 3. Define Week 1 Learning Log Entries

Create a list of dictionaries containing the day, topic, and reflection notes for each Week 1 learning entry.

```python
# Define Week 1 learning log entries
learning_entries = [
    {"day": "Day 1", "topic": "Python Introduction & Hello World", 
     "reflection": "Learned how to use print() to display output in Python. Created my first Hello World program."},
    {"day": "Day 2", "topic": "Variables and Data Types", 
     "reflection": "Learned about int, float, str, and bool data types. Practiced declaring and printing variables of each type."},
    {"day": "Day 3", "topic": "Basic Arithmetic Operations", 
     "reflection": "Practiced addition, subtraction, multiplication, and division. Used variables to perform calculations."},
    {"day": "Day 4", "topic": "String Concatenation and f-Strings", 
     "reflection": "Learned string concatenation using the + operator and modern f-strings for cleaner string formatting."},
    {"day": "Day 5", "topic": "User Input and Type Conversion", 
     "reflection": "Used input() to get user data, converted strings to integers, and practiced type checking with type()."}
]

print("Learning entries defined successfully!")
print(f"Total entries: {len(learning_entries)}")
```

## 4. Create the Output Folder

Use `pathlib` to create a folder named `week1_learning_log` inside the working directory if it does not already exist.

```python
# Create the output folder
output_dir = Path("week1_learning_log")
output_dir.mkdir(exist_ok=True)
print(f"Folder '{output_dir}' is ready at: {output_dir.resolve()}")
```

## 5. Write Entries to a Markdown File

Iterate over the learning log entries and write them into a formatted markdown file saved inside the `week1_learning_log` folder.

```python
# Write entries to a Markdown file
md_file = output_dir / "week1_learning_log.md"

with open(md_file, "w", encoding="utf-8") as f:
    f.write("# Week 1 Learning Log\n\n")
    f.write(f"**Generated on:** {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n\n")
    f.write("---\n\n")
    
    for entry in learning_entries:
        f.write(f"## {entry['day']}: {entry['topic']}\n\n")
        f.write(f"**Reflection:** {entry['reflection']}\n\n")
        f.write("---\n\n")

print(f"Markdown file created at: {md_file.resolve()}")
print(f"File size: {md_file.stat().st_size} bytes")
```

## 6. Save the Log as a CSV File

Convert the learning log entries into CSV format and write the data to a file inside the same folder.

```python
# Save the log as a CSV file
csv_file = output_dir / "week1_learning_log.csv"

fieldnames = ["day", "topic", "reflection"]
with open(csv_file, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(learning_entries)

print(f"CSV file created at: {csv_file.resolve()}")
print(f"File size: {csv_file.stat().st_size} bytes")
```

## 7. Verify the Folder Contents

List all files in the folder to confirm that the markdown and CSV files were created correctly.

```python
# Verify the folder contents
print(f"Contents of '{output_dir}':")
for file_path in sorted(output_dir.iterdir()):
    print(f"  - {file_path.name} ({file_path.stat().st_size} bytes)")

print("\n✅ Learning log created successfully!")
```
